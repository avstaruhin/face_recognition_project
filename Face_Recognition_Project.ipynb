# Обучение нейросети для распознавания лиц на датасете CelebA-500

Что такое "распознавание": мы хотим построить такую нейросеть, которая принимает на вход картинку лица, а выдает вектор определенного размера. И мы хотим, чтобы векторы лиц одного и того же человека были близки по некоторому расстоянию (например, косинусному), а векторы лиц разных людей были далеки по этому же расстоянию. Обычно в качестве расстояния используется косинусное расстояние (cosine distance).

Почему мы хотим построить именно такую сеть? Почему не посмтроить обычную сеть для классификации и научить ее классифицировать лица из обучающей выборки, где каждый отдельный человек — это один класс? Ответ такой — мы хотим, чтобы наша модель умела работать с лицами людей, которых нет в обучающей выборке. Хотим, чтобы она на лица одного и того же незнакомого человека выдавала близкие векторы (т.е. "говорила" нам, что это один и тот же человек), а на лица разных людей — далекие векторы.

Пусть тогда у нас есть база данных людей вида (id, набор векторов его лиц). Когда к нам придет новая картинка лица, мы ее прогоним через модель и получим вектор этого лица. Далее сравним этот вектор по метрике расстояния с векторами из нашей базы. Тот id, чьи векторы будут наиболее близки к нашему, и будет ответом на вопрос "кто изображен на картинке"

*Создадим 3 датасета: тренировочный, валидационный, тестовый*

```python
# Путь к папке с изображениями
folder_path = '/content/data_dls_face_recognition/celebA_train_500/celebA_imgs/'

# Пути к файлам с аннотациями и разделением данных
anno_file_path = '/content/data_dls_face_recognition/celebA_train_500/celebA_anno.txt'
split_file_path = '/content/data_dls_face_recognition/celebA_train_500/celebA_train_split.txt'

# Преобразования для изображений
transform = transforms.Compose([
    transforms.Lambda(lambda image: np.array(image)[77:-41, 45:-50]),  # Кропаем каждое изображение
    transforms.ToPILImage(),  # Преобразуем обратно в объект JpegImageFile
    transforms.Resize((160, 160)),  # Изменяем размер изображения, растягиваем в размер 160 х 160 px
    transforms.ToTensor(), # Преобразуем изображение в тензор Torch
])

# Чтение меток классов из файла аннотаций
annotations = {}
with open(anno_file_path, 'r') as file:
    for line in file:
        parts = line.split()
        if len(parts) == 2:
            image_name, label = parts
            annotations[image_name] = int(label)

# Чтение меток разделения на train/validation/test из файла
labels_and_splits = {}
with open(split_file_path, 'r') as file:
    for line in file:
        parts = line.split()
        if len(parts) == 2:
            image_name, split = parts
            labels_and_splits[image_name] = int(split)

# Разделение данных на три набора в зависимости от меток классов (0, 1, 2) (train/validation/test соответственно)
train_dataset = [(image_name, annotations[image_name]) for image_name in labels_and_splits if labels_and_splits[image_name] == 0]
valid_dataset = [(image_name, annotations[image_name]) for image_name in labels_and_splits if labels_and_splits[image_name] == 1]
test_dataset = [(image_name, annotations[image_name]) for image_name in labels_and_splits if labels_and_splits[image_name] == 2]

# Создание объекта Dataset
class CustomDataset(torch.utils.data.Dataset):
    def __init__(self, dataset, folder_path, transform=None):
        self.dataset = dataset
        self.folder_path = folder_path
        self.transform = transform

    def __len__(self):
        return len(self.dataset)

    def __getitem__(self, idx):
        image_name, label = self.dataset[idx]
        image_path = os.path.join(self.folder_path, image_name)

        # Открываем изображение
        img = Image.open(image_path)

        # Применяем преобразования, если они указаны (в нашем случае кропаем, растягиваем, преобразуем к тензору)
        if self.transform:
            img = self.transform(img)

        return img, label

# Создание объекта Dataset для каждого разделения (train/validation/test)

# Тренировочный датасет
train_dataset = CustomDataset(train_dataset, folder_path, transform=transform)
# Валидационный датасет
valid_dataset = CustomDataset(valid_dataset, folder_path, transform=transform)
# Тестовый датасет
test_dataset = CustomDataset(test_dataset, folder_path, transform=transform)
```

*Функция дообучения предобученной модели*

```python
# Функция обучения
def fine_tune_efficientnetv3(model, train_loader, val_loader, criterion, optimizer, scheduler, batch_size, num_epochs=5):

    """
    input:
      model: модель, дообучающаяся на данных
      train_loader: данные для обучения
      val_loader: данные для валидации
      criterion: критерий (функция потерь)
      optimizer: оптимизатор
      scheduler: планировщик, контролирующий шаг оптимизации (learning rate)
      batch_size: размер бача
      num_epochs: количество эпох обучения (по умолчанию: 5)
    output:
      history: исторические данные обучения
    """

    # Создаем DataLoader для обучающего набора данных с перемешиванием
    train_loader = DataLoader(train_loader, batch_size=batch_size, shuffle=True)
    # Создаем DataLoader для валидационного набора данных, но без перемешивания
    val_loader = DataLoader(val_loader, batch_size=batch_size, shuffle=False)

    # Определяем устройство (GPU или CPU) для выполнения вычислений
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    # Переносим модель на определенное устройство (GPU или CPU)
    model.to(device)

    # Создаем словарь для отслеживания значений потерь и точности на обучающем и валидационном наборах данных
    history = {'train_loss': [], 'val_loss': [], 'train_accuracy': [], 'val_accuracy': []}

    # Начинаем цикл обучения по заданному числу эпох
    for epoch in range(num_epochs):

        # Начало эпохи тренировки
        start_train = time.time()
        # Установление модели в режим обучения
        model.train()

        # Обнуляем переменные
        running_loss = 0.0
        correct_train = 0
        total_train = 0

        # Переносим входные данные и метки на устройство (GPU или CPU)
        for inputs, labels in train_loader:
            inputs, labels = inputs.to(device), labels.to(device)

            # Обнуляем градиенты перед обратным распространением
            optimizer.zero_grad()

            # Получаем прогнозы модели для входных данных
            outputs = model(inputs)
            # Вычисляем потери между прогнозами и истинными метками
            loss = criterion(outputs, labels)
            # Выполняем обратное распространение для вычисления градиентов
            loss.backward()
            # Обновляем веса модели на основе градиентов
            optimizer.step()

            # Накапливаем потери на текущем батче
            running_loss += loss.item()

            # Находим индексы максимальных значений в выходах модели
            _, predicted = outputs.max(1)
            # Увеличиваем общее количество обучающих образцов на размер текущего батча
            total_train += labels.size(0)
            # Увеличиваем количество правильно предсказанных образцов за текущий батч
            correct_train += predicted.eq(labels).sum().item()

        # Вычисляем средние потери за эпоху
        epoch_loss = running_loss / len(train_loader)
        # Вычисляем среднюю точность за эпоху
        epoch_accuracy = correct_train / total_train

        # Добавляем значение средних потерь за эпоху в историю обучения
        history['train_loss'].append(epoch_loss)
        # Добавляем значение средней точности за эпоху в историю обучения
        history['train_accuracy'].append(epoch_accuracy)

        # Окончание эпохи тренировки
        end_train = time.time()

        # Выводим информацию о средних потерях и точности для текущей эпохи
        print(f"Epoch {epoch+1}/{num_epochs}, Train Loss: {epoch_loss}, Train Accuracy: {epoch_accuracy}, Time: {end_train - start_train}")

        # Обновление learning rate с использованием планировщика
        scheduler.step()

        # Валидация на каждой эпохе (если есть валидационный DataLoader)
        if val_loader is not None:

            # Начало эпохи валидации
            start_val = time.time()

            # Установление модели в режим оценки (валидации)
            model.eval()

            # Обнуляем переменные
            running_val_loss = 0.0
            correct_val = 0
            total_val = 0

            # Блок кода, в котором градиенты вычисляться не будут
            with torch.no_grad():
                # Этот цикл проходит через валидационный DataLoader, который предоставляет входные данные и соответствующие метки
                for inputs_val, labels_val in val_loader:
                    # Переносим входные данные и метки на устройство (GPU или CPU), в зависимости от доступности GPU
                    inputs_val, labels_val = inputs_val.to(device), labels_val.to(device)

                    # Получаем выходы (предсказания) модели для валидационных данных
                    outputs_val = model(inputs_val)
                    # Вычисляем потери между предсказаниями и истинными метками
                    val_loss = criterion(outputs_val, labels_val)

                    # Накапливаем общие потери на протяжении всей валидации
                    running_val_loss += val_loss.item()

                    # Вычисляем индексы классов с максимальной вероятностью (предсказанные классы)
                    _, predicted_val = outputs_val.max(1)
                    # Увеличиваем общее количество валидационных образцов
                    total_val += labels_val.size(0)
                    # Увеличиваем количество правильных предсказаний, сравнивая предсказанные классы с истинными метками
                    correct_val += predicted_val.eq(labels_val).sum().item()

            # Вычисляем средние потери на валидационном наборе данных
            epoch_val_loss = running_val_loss / len(val_loader)
            # Вычисляем точность на валидационном наборе данных
            epoch_val_accuracy = correct_val / total_val

            # Добавляем средние потери в историю валидационных потерь для последующего анализа
            history['val_loss'].append(epoch_val_loss)
            # Добавляем точность в историю валидационной точности для последующего анализа
            history['val_accuracy'].append(epoch_val_accuracy)

            # Окончание эпохи валидации
            end_val = time.time()

            # Выводим информацию о потерях и точности на валидационном наборе данных после каждой эпохи обучения
            print(f"Epoch {epoch+1}/{num_epochs}, Validation Loss: {epoch_val_loss}, Validation Accuracy: {epoch_val_accuracy}, Time: {end_val - start_val}")

    # Выводим сообщение о завершении обучения
    print("Training complete")

    # Возвращаем словарь с историей значений потерь и точности для обучающего и валидационного наборов данных
    return history
```

*Функция предсказания модели*

```python
# Функция предсказания
def predict(model, test_loader, criterion, batch_size):

    """
    input:
      model: модель для валидации
      test_loader: набор данных для проверки качества и потерь модели
      criterion: критерий для вычисления функции потерь
      batch_size: размер бача
    output:
      None
    """

     # Установление модели в режим оценки (тестирования)
    model.eval()

    # Обнуляем переменные
    running_test_loss = 0.0
    correct_test = 0
    total_test = 0

     # Создаем DataLoader для тестового набора данных без перемешивания
    test_loader = DataLoader(test_loader, batch_size=batch_size, shuffle=False)

    # Блок кода, в котором градиенты вычисляться не буду
    with torch.no_grad():
        # Этот цикл проходит через тестовый DataLoader, который предоставляет входные данные и соответствующие метки
        for inputs_test, labels_test in test_loader:
            # Переносим входные данные и метки на устройство (GPU или CPU), в зависимости от доступности GPU
            inputs_test, labels_test = inputs_test.to(device), labels_test.to(device)

            # Получаем выходы (предсказания) модели для тестовых данных
            outputs_test = model(inputs_test)
            # Вычисляем потери между предсказаниями и истинными метками
            test_loss = criterion(outputs_test, labels_test)

            # Накапливаем общие потери на протяжении всего тестирования
            running_test_loss += test_loss.item()

            # Вычисляем индексы классов с максимальной вероятностью (предсказанные классы)
            _, predicted_test = outputs_test.max(1)
            # Увеличиваем общее количество тестовых образцов
            total_test += labels_test.size(0)
            # Увеличиваем количество правильных предсказаний, сравнивая предсказанные классы с истинными метками
            correct_test += predicted_test.eq(labels_test).sum().item()

    # Вычисляем средние потери на тестовом наборе данных
    epoch_test_loss = running_test_loss / len(test_loader)
    # Вычисляем точность на тестовом наборе данных
    epoch_test_accuracy = correct_test / total_test

    # Вывод информации
    print(f'Loss на тестовых данных: {epoch_test_loss}')
    print(f'Accuracy на тестовых данных: {epoch_test_accuracy}')
```

# Реализация Identification Rate Metric

При обучении модели для распознавания лиц с помощью CE (кросс-энтропии) мы можем считать метрику accuracy как индикатор того, насколько хорошо наша модель работает. Но у accuracy тут есть недостаток: она не сможет померить, насколько хорошо наша модель работает на лицах людей, которых нет в обучающей выборке.

Чтобы это исправить, придумали новую метрику: identification rate. Вот как она работает:

Создадим два набора изображений лиц: query и distractors. Никакие лица из этих наборов не должны содержаться в обучающем и валидационном датасете.

1. посчитаем косинусные расстояния между лицами, соответствующими одним и тем же людям из query части. Например, пусть одному человеку соответствуют три фото в query: 01.jpg, 02.jpg, 03.jpg. Тогда считаем три косинусных расстояния между всеми тремя парами из этих фото.
2. посчитаем косинусные расстояния между лицами, соответствующими разным людям из query части.
3. посчитаем косинусные расстояния между всеми парами лиц из query и distractors. Т.е. пара — это (лицо из query, лицо из distractors). Всего получится |query|*|distractors| пар.
4. Сложим количества пар, полученных на 2 и 3 шагах. Это количество false пар.
5. Зафиксируем FPR (false positive rate). Пусть, например, будет 0.01. FPR, умноженный на количество false пар из шага 4 — это разрешенное количество false positives, которые мы разрешаем нашей модели. Обозначим это количество через N.
6. Отсортируем все значения косинусных расстояний false пар. N — ое по счету значение расстояния зафиксируем как пороговое расстояние.
7. Посчитаем количество positive пар с шага 1, которые имеют косинусное расстояние меньше, чем пороговое расстояние. Поделим это количество на общее количество positive пар с шага 1. Это будет TPR (true positive rate) — итоговое значение нашей метрики.

Такая метрика обычно обозначается как TPR@FPR=0.01. FPR может быть разным. Приразных FPR будет получаться разное TPR.

Смысл этой метрики в том, что мы фиксируем вероятность ошибки вида false positive, т.е. когда "сеть сказала, что это один и тот же человек, но это не так", считаем порог косинусного расстояния для этого значения ошибки, потом берем все positive пары и смотрим, у скольких из них расстояние меньше этого порога. Т.е. насколько точно наша сеть ищет похожие лица при заданной вероятности ошибки вида false positive.

```python
# Загрузка изображения из файла
def load_image_as_tensor(image_path):

    """
    Загрузка изображения в тензор с применением преобразований
    input:
      image_path: путь к изображению
    output:
      tensor_image: полученный тензор изображения
    """

    # Открывает изображение из файла по указанному пути с использованием библиотеки PIL
    image = Image.open(image_path).convert("RGB")

    # Применяем преобразования (если они предоставлены)
    transform = transforms.Compose([
    transforms.Resize((160, 160)),  # Изменяем размер изображения, растягиваем в размер 160 х 160
      ])

    if transform:
        image = transform(image)

    # Преобразуем изображение в тензор
    tensor_image = transforms.ToTensor()(image)

    # Добавляем размерность батча
    tensor_image = tensor_image.unsqueeze(0)

    # Возвращает полученный тензор
    return tensor_image

# Функция, возвращающая список эмбеддингов для всех изображений из входного списка
def compute_embeddings(model, images_list, mode='query'):

  '''
  compute embeddings from the trained model for list of images.
  params:
    model: trained nn model that takes images and outputs embeddings
    images_list: list of images paths to compute embeddings for
  output:
    list: list of model embeddings. Each embedding corresponds to images
          names from images_list
  '''

  # Создает пустой список для хранения эмбеддингов
  embeddings = []

  model.eval()  # Установка модели в режим оценки (не обучения)

  #  Создает контекст, в котором отключено вычисление градиентов
  with torch.no_grad():
      # Итерирует по списку путей к изображениям
      for image_path in images_list:
          # Загрузка изображения в тензор без предварительной обработки
          # Разделение на режимы в зависимости от query или distractors
          if mode == 'query':
            # Путь к папке с изображениями query
            image_path = '/content/data_dls_face_recognition/celebA_ir/celebA_query/' + image_path

          if mode == 'distractors':
            # Путь к папке с изображениями distractors
            image_path = '/content/data_dls_face_recognition/celebA_ir/celebA_distractors/' + image_path

          # Загружает изображение и преобразует его в тензор
          image_tensor = load_image_as_tensor(image_path)
          # Переносит тензор на устройство
          image_tensor = image_tensor.to(device)
          # Получает выходы с предпоследнего слоя для данного изображения
          dropout_layer_output = get_prelast_layer_output(efficientnetv3, image_tensor)
          # Преобразование тензора в список
          embedding_list = dropout_layer_output.squeeze().tolist()
          # Добавляет полученный список в общий список эмбеддингов
          embeddings.append(embedding_list)

  # Возвращает список эмбеддингов для всех изображений из входного списка
  return embeddings

  # YOUR CODE HERE
```

```python
# Функция нахождение выходов предпоследнего слоя
def get_prelast_layer_output(model, input_data):

    """
    input:
      model: модель, с которой требуется снять выходы
      input_data: данные, выходы которых требуется снять
    output:
      prelast_layer_output: выходы предпоследнего слоя модели
    """

    # Создаем переменную для хранения выходов предпоследнего слоя
    prelast_layer_output = None

    # Функция, которая будет вызываться при прохождении данных через предпоследний слой
    def hook(module, input, output):
        nonlocal prelast_layer_output
        prelast_layer_output = output

    # Находим предпоследний слой в модели (второй с конца)
    prelast_layer = list(model.children())[-3]

    # Регистрируем хук на предпоследнем слое
    hook_handle = prelast_layer.register_forward_hook(hook)

    # Пропускаем данные через модель
    model(input_data)

    # Удаляем хук после использования
    hook_handle.remove()

    # Возвращает выходы предпоследнего слоя модели на данные
    return prelast_layer_output
```

```python
# Функция вычисляет косинусное сходство между парами векторов для изображений query
def compute_cosine_query_pos(query_dict, query_img_names, query_embeddings):

    """
    input:
      query_dict: словарь, содержащий информацию о классах изображений
      query_img_names: список имен изображений
      query_embeddings: список эмбеддингов для всех изображений query
    output:
      similarities: список косинусных сходств между парами векторов
    """

    # Создает пустой список для хранения косинусных сходств между парами векторов
    similarities = []

    # Проходится по значениям словаря query_dict, содержащего информацию о классах изображений
    for class_images in query_dict.values():
        # Внутренний цикл по индексам изображений в текущем классе, начиная с первого
        for i in range(len(class_images) - 1):
            # Внутренний цикл по индексам изображений в текущем классе, начиная с индекса следующего после i
            for j in range(i + 1, len(class_images)):

                # Находит индекс изображения i в списке имен изображений.
                index_i = query_img_names.index(class_images[i])
                #  Находит индекс изображения j в списке имен изображений
                index_j = query_img_names.index(class_images[j])

                # Получает вектор встроенных представлений для изображения i и преобразует его в массив NumPy
                embedding_i = np.array(query_embeddings[index_i]).reshape(1, -1)
                # Аналогично получает вектор встроенных представлений для изображения j и преобразует его
                embedding_j = np.array(query_embeddings[index_j]).reshape(1, -1)

                # Вычисляет косинусное сходство между векторами embedding_i и embedding_j с помощью функции cosine_similarity
                similarity = cosine_similarity(embedding_i, embedding_j)[0][0]
                # Добавление результата в список
                similarities.append(similarity)

    # Возвращает список косинусных сходств между парами векторов встроенных представлений
    return similarities

# Функция, возвращающая список косинусных сходств между парами векторов встроенных представлений для изображений разных классов
def compute_cosine_query_neg(query_dict, query_img_names, query_embeddings):

    """
    input:
      query_dict: словарь, содержащий информацию о классах изображений
      query_img_names: список имен изображений
      query_embeddings: список эмбеддингов для всех изображений query
    output:
      similarities: список косинусных сходств между парами векторов
    """

    # Создает пустой список для хранения косинусных сходств между парами векторов
    similarities = []

    # Внешний цикл по индексам изображений в запросе, начиная с первого
    for i in range(len(query_img_names) - 1):
        # Внутренний цикл по индексам изображений в запросе, начиная с индекса следующего после i
        for j in range(i + 1, len(query_img_names)):

            # Находит класс (группу) изображения i в словаре query_dict, используя next для поиска первого значения, удовлетворяющего условию
            class_i = next((key for key, value in query_dict.items() if query_img_names[i] in value), None)
            # Аналогично находит класс изображения j
            class_j = next((key for key, value in query_dict.items() if query_img_names[j] in value), None)
            # Проверяет, принадлежат ли изображения i и j к разным классам
            if class_i != class_j:
                # Получает вектор встроенных представлений для изображения i
                embedding_i = np.array(query_embeddings[i]).reshape(1, -1)
                # Аналогично получает вектор встроенных представлений для изображения j и преобразует его
                embedding_j = np.array(query_embeddings[j]).reshape(1, -1)

                # Вычисляет косинусное сходство между векторами embedding_i и embedding_j с помощью функции cosine_similarity
                similarity = cosine_similarity(embedding_i, embedding_j)[0][0]
                # Добавление результата в список
                similarities.append(similarity)

    # Возвращает список косинусных сходств между парами векторов встроенных представлений для изображений разных классов
    return similarities

# Функция вычисляет косинусное сходство между векторами изображений query_embeddings и векторами distractors_embeddings
def compute_cosine_query_distractors(query_embeddings, distractors_embeddings):

    """
    input:
      query_embeddings: список эмбеддингов для всех изображений query
      distractors_embeddings: список эмбеддингов для всех изображений distractors
    output:
      similarities: список косинусных сходств между парами векторов
    """

    # Создает пустой список для хранения косинусных сходств между векторами встроенных представлений
    similarities = []

    # Внешний цикл по векторам встроенных представлений изображений запроса
    for query_embedding in query_embeddings:
        # Внутренний цикл по векторам встроенных представлений distractor_embeddings
        for distractor_embedding in distractors_embeddings:
            # Преобразует текущий вектор встроенных представлений запроса в массив NumPy
            query_embedding = np.array(query_embedding).reshape(1, -1)
            # Преобразует текущий вектор встроенных представлений дистрактора в массив NumPy
            distractor_embedding = np.array(distractor_embedding).reshape(1, -1)

            # Вычисляет косинусное сходство между текущим вектором query и вектором distractor
            similarity = cosine_similarity(query_embedding, distractor_embedding)[0][0]
            # Результат добавляется в список
            similarities.append(similarity)

    # Возвращает список косинусных сходств между векторами встроенных представлений изображений query и distractor
    return similarities
```

# Реализация ArcFace loss

**Как устроен ArcFace**:

Стандартные SoftMax + кросс-энтропия (CE) выглядят так:

$$L_{CE} = \frac{-1}{N}\sum_1^N \frac{e^{W_{y_i}^{T}x_i + b_{y_i}}}{\sum^n_{j=1}e^{W_j^Tx_i+b_j}},$$

здесь:
- $x_i \in \mathbb{R^d}$ — вектор $i$-го элемента обучающей выборки перед последним полносвязным слоем сети. $y_i$ — класс этого элемента;
- $W_j \in \mathbb{R^d}$ — j-ый столбец матрицы весов последнего слоя сети (т.е. слоя, который производит итоговую классификацю входящего объекта);
- $b_j \in \mathbb{R^d}$ — j-ый элемент вектора байеса последнего слоя сети;
- $N$ — batch size;
- $n$ — количество классов.


Хотя этот лосс работает хорошо, он явным образом не заставляет эмбеддинги $x_i$ элементов, принадлежащих одному классу, быть близкими друг к другу по расстоянию. И не заставляет эмбеддинги элементов, принадлежащих разным классам, быть далеко друг от друга. Все, что хочет этот лосс — чтобы на основе эмбеддингов $x_i$ можно было хорошо классифицировать элементы, никакие ограничений на расстояния между эмбеддингами $x_i$ он не вводит.

Из-за этого у нейросетей для распознавания лиц, которые обучены на обычном CE loss, бывают проблемы с распознаванием лиц, которые сильно отличаются от лиц того же человека разными допатрибутами (шляпа/прическа/очки и т.п.). Просто эмбеддинг для таких лиц получается довольно далек по расстоянию от других эмбеддингов лиц этого же человека.

Давайте теперь немного поправим формулу:
- уберем байес последнего слоя, т.е. сделаем $b_j=0$;
- нормализуем веса последнего слоя: ||$W_j$|| = 1;
- нормализуем эмбеддинги: ||$x_i$|| = 1. Перед подачей их на вход последнему слою (т.е. перед умножением на матрицу $W_j$) умножим их на гиперпараметр s. По сути, мы приводим норму всех эмбеддингов к s. Смысл этого гиперпараметра в том, что, возможно, сети проще будет классифицировать эмбеддинги, у которых не единичная норма.

Нормализация эмбеддингов приводит к тому, что эмбеддинги начинают быть распределены по сфере единичного радиуса (и сфере радиуса s после умножения на ниперпараметр s). И итоговые предсказания сети после последнего слоя зависят только от угла между эмбеддингами $x_i$ и выученных весов $W_j$. От нормы эмбеддинга $x_i$ они больше не зависят, т.к. у всех эмбеддингов они теперь одинаковые.

Получается, в степени экспоненты у нас останется выражение $s W_{y_i}^{T}x_i$, которое можно переписать в виде  $s W_{y_i}^{T}x_i = s ||W_{y_i}||\cdot ||x_i|| \cdot cos\Theta_{y_i}$. Тут $\Theta_{y_i}$ — это угод между векторами $W_{y_i}$ и $x_i$. Но так как мы сделали нормы $W_{y_i}$ и $x_i$ единичными, то все это выражение просто будет равно $s cos\Theta_{y_i}$.

В итоге мы получим следующую формулу лосса:

$$L = \frac{-1}{N}\sum_1^N \frac{e^{s\ cos\Theta_{y_i}}}{e^{s\ cos\Theta_{y_i}} + \sum^n_{j=1,\ j\ne y_i} e^{s\ cos\Theta_j}}$$


И последний шаг. Добавим еще один гиперпараметр $m$. Он называется additive angular margin penalty и заставляет эмбеддинги одного класса быть ближе друг к другу, а эмбеддинги разных классов — более далекими друг от друга.

В итоге получим вот что:

$$L_{ArcFace} = \frac{-1}{N}\sum_1^N \frac{e^{s\ cos(\Theta_{y_i} + m)}}{e^{s\ cos(\Theta_{y_i} + m)} + \sum^n_{j=1,\ j\ne y_i} e^{s\ cos\Theta_j}}$$

Это и есть ArcFace Loss с двумя  гиперпараметрами, s и m.

Получается, что ArcFace Loss завтавляет сеть выучивать эмбеддинги, распределенные по сфере радиуса s, причем чтобы эмбеддинги одного класса были ближе друг к другу, а эмбеддинги разных классов — более далеки друг от друга.

```python
# Функция вычисляет точность (accuracy) и потери (loss) для модели на заданном наборе данных
def compute_accuracy(model, data_loader, device):

    """
    input:
      model: модель, для которой требуется вычислить accuracy и loss
      data_loader: данные, на которых валидируется модель
      device: устройство
    output:
      accuracy: точность предсказаний модели
      loss: потери модели
    """

    # Устанавливает модель в режим оценки
    model.eval()
    # Инициализируют счетчики для правильных предсказаний и общего количества образцов
    correct_predictions = 0
    total_samples = 0

    # Запускает блок кода с отключенным трекингом градиентов
    with torch.no_grad():
        # Проходится по загрузчику данных
        for inputs, labels in data_loader:
            # Переносит входные данные и метки на устройство
            inputs, labels = inputs.to(device), labels.to(device)

            # Проходит входные данные через модель для получения выходных значений
            outputs = model(inputs)
            # Вычисляет предсказанные классы, выбирая класс с максимальным значением из выходов
            _, predicted = torch.max(outputs, 1)
            #  Увеличивает счетчик правильных предсказаний на количество верных предсказаний в текущей итерации
            correct_predictions += (predicted == labels).sum().item()
            # Увеличивает общее количество образцов на количество образцов в текущей итерации
            total_samples += labels.size(0)

    # Вычисляет точность как отношение верных предсказаний к общему количеству образцов
    accuracy = correct_predictions / total_samples
    # Вычисление потерь
    loss = arcface_loss(outputs, labels)

    # Возвращает точность предсказаний и потери
    return accuracy, loss
```

Определение функции потерь Arc Face Loss

```python
# Определение функции потерь (ArcFace Loss)
class ArcFaceLoss(nn.Module):
    def __init__(self, s=5.0, m=0.5):
        super(ArcFaceLoss, self).__init__()
        self.s = s  # Гиперпараметр s
        self.m = m  # Гиперпараметр m

    def forward(self, logits, labels):

        """
        input:
          logits: предсказанные логиты
          labels: метки классов
        output:
          loss: вычисленные потери Arc Face
        """

        # Нормализуем веса перед подсчетом cos(theta)
        logits = logits / torch.norm(logits, dim=1, keepdim=True)

        # Вычисляем cos(theta) для соответствующих классов
        cos_theta = logits.gather(1, labels.view(-1, 1))

        # Вычисляем theta
        theta = torch.acos(cos_theta)

        # Добавляем additive angular margin penalty
        target_theta = theta + self.m
        target_cos_theta = torch.cos(target_theta)

        # Вычисляем logit для соответствующего угла
        logits = logits - self.s * (target_cos_theta - cos_theta)

        # Вычисляем Arc Face Loss
        loss = nn.CrossEntropyLoss()(logits, labels)

        # Возвращаем Arc Face Loss
        return loss
```

# Реализация Triplet Loss

Снова скажем, что идея лосса — заставить эмбеддинги лиц одного человека быть более близкими по некоторому расстоянию, а эмбеддинги лиц разных людей — далекими друг от друга. Общая формула лосса выглядит так:

$$L(e, p, n) = max\{d(a, p) - d(a, n) + margin, 0\},$$

здесь
- $e$ — эмбеддинг входного лица (output модели)
- $p$ — "positive" эмбеддинг для входного лица (т.е. эмбеддинг такого элемента, что мы хотим, чтобы $e$ и $p$ были близки. В нашем случае это значит, что $e$ и $p$ должны быть выходами сети на два разных фото одного и того же человека).
- $n$ — "negative" эмбеддинг для входного лица (т.е. эмбеддинг такого элемента, что мы хотим, чтобы $e$ и $p$ были далеки. В нашем случае это значит, что $e$ и $p$ должны быть выходами сети на два разных фото разных людей).
- $d(x, y)$ — метрика расстояния между эмбеддингами, по которой мы их сравниваем.
- margin — гиперпараметр, который заставляет $d(a, p)$ и $d(a, n)$ быть еще дальше друг от друга.

**Эмбеддинги $e$, $p$ и $n$ нужно нормализовать, прежде чем подавать в лосс-функцию**.

У TripletLoss есть куча разных вариаций. В некоторых из них больше гиперпараметров, в других предлагают использовать больше одного позитивного и негативного эмбеддинга за раз. Где-то предлагается умный способ выбора negative эмбеддинга (например, выбирается такой, на котором нейросеть пока плохо работает, т.е. считает $e$ и $n$ близкими).

```python
# Определение класса TripletDataset
class TripletDataset(Dataset):
    # Инициализация
    def __init__(self, folder_path, anno_file_path, split_file_path, split_type=0, transform=None):
        # Путь к папке с изображениями
        self.folder_path = folder_path
        # Преобразование изображений
        self.transform = transform
        # Тип разделения (по умолчанию 0)
        self.split_type = split_type
        # Путь к файлу с аннотациями
        self.annotations = self.load_annotations(anno_file_path)
        # Путь к файлу с разделением (train, val, test)
        self.split_labels = self.load_split_labels(split_file_path)

        # Фильтруем аннотации в соответствии с разделением
        self.filtered_annotations = {image_name: label for image_name, label in self.annotations.items()
                                     if self.split_labels[image_name] == split_type}
        # Фильтрация по встречаемости меток
        value_counts = Counter(self.filtered_annotations.values())

        # Отфильтровываем значения, которые встречаются больше одного раза так,
        # как к одному изображению класса нельзя будет найти позитивное в датасете
        self.filtered_annotations = {image_name: label for image_name, label in self.filtered_annotations.items()
                              if value_counts[label] > 1}

        value_counts = Counter(self.filtered_annotations.values())

    # Функция загрузки аннотаций
    def load_annotations(self, file_path):
        annotations = {}
        with open(file_path, 'r') as file:
            for line in file:
                parts = line.split()
                if len(parts) == 2:
                    image_name, label = parts
                    annotations[image_name] = int(label)

        return annotations

    # Загрузка информации о разделении данных
    def load_split_labels(self, file_path):
        split_labels = {}
        with open(file_path, 'r') as file:
            for line in file:
                parts = line.split()
                if len(parts) == 2:
                    image_name, split_label = parts
                    split_labels[image_name] = int(split_label)

        return split_labels

    def __len__(self):
        return len(self.filtered_annotations)

    def __getitem__(self, anchor_idx):
        # Определение якоря (anchor)
        anchor_image_name = list(self.filtered_annotations.keys())[anchor_idx]
        anchor_label = self.filtered_annotations[anchor_image_name]

        # Выбираем случайный положительный пример из того же класса, что и якорь (anchor)
        positive_candidates = [image_name for image_name, label in self.filtered_annotations.items()
                                if label == anchor_label and image_name != anchor_image_name]
        positive_image_name = random.choice(positive_candidates)

        # Выбираем случайный отрицательный пример из другого класса
        negative_candidates = [image_name for image_name, label in self.filtered_annotations.items()
                                if label != anchor_label]
        negative_image_name = random.choice(negative_candidates)

        # Загружаем изображения
        anchor_img = self.load_image(anchor_image_name)
        positive_img = self.load_image(positive_image_name)
        negative_img = self.load_image(negative_image_name)

        # Возвращаем anchor, positive, negative
        return anchor_img, positive_img, negative_img

    # Загрузка изображения
    def load_image(self, image_name):

        """
        input:
          self: экземпляр класса
          image_name: имя изображения
        output:
          img: преобразованное изображение
        """

        image_path = os.path.join(self.folder_path, image_name)
        img = Image.open(image_path).convert('RGB')
        if self.transform:
            img = self.transform(img)
        return img

# Преобразования для изображений
transform = transforms.Compose([
    transforms.Lambda(lambda image: np.array(image)[77:-41, 45:-50]),  # Кропаем каждое изображение
    transforms.ToPILImage(),  # Преобразуем обратно в объект JpegImageFile
    transforms.Resize((160, 160)),  # Изменяем размер изображения, растягиваем в размер 160 х 160 px
    transforms.ToTensor(), # Преобразуем изображение в тензор Torch
])

# Пути к файлам
folder_path = '/content/data_dls_face_recognition/celebA_train_500/celebA_imgs/'
anno_file_path = '/content/data_dls_face_recognition/celebA_train_500/celebA_anno.txt'
split_file_path = '/content/data_dls_face_recognition/celebA_train_500/celebA_train_split.txt'

# Создание датасетов
triplet_train_dataset = TripletDataset(folder_path, anno_file_path, split_file_path, split_type=0, transform=transform)
triplet_valid_dataset = TripletDataset(folder_path, anno_file_path, split_file_path, split_type=1, transform=transform)
triplet_test_dataset = TripletDataset(folder_path, anno_file_path, split_file_path, split_type=2, transform=transform)
```

Напишим функцию измерения качества для Triplet Loss

Смысл в следующем: будем считать, что модель угадала, если расстояние между якорем и позитивом будет меньше, чем расстояние между якорем и негативом, умноженное на пороговое значение

В итоге посчитаем точность разделив количество угаданных изображений на общее количество данных в датасете

```python
# Функция качества модели
def triplet_accuracy(anchor_embedding, positive_embedding, negative_embedding, threshold=0.5):

    """
    input:
      anchor_embedding: вектора эмбеддингов якоря
      positive_embedding: вектора эмбеддингов позитивного изображения
      negative_embedding: вектора эмбеддингов негативного изображения
      threshold: порог (по умолчанию: 0.5)
    output:
      accuracy: точность предсказаний модели
    """

    # Вычисление расстояний между векторами эмбеддингов
    distance_positive = F.pairwise_distance(anchor_embedding, positive_embedding, p=2)
    distance_negative = F.pairwise_distance(anchor_embedding, negative_embedding, p=2)

    # Вычисление количества правильных триплетов
    correct_triplets = (distance_positive < threshold * distance_negative).sum().item()
    # Вычисление общего числа триплетов
    total_triplets = anchor_embedding.size(0)

    # Вычисление точности
    accuracy = correct_triplets / total_triplets

    # Возврат точности
    return accuracy
```

```python
# Функция обучения модели на Triplet Loss
def train_and_validate_triplet_loss(model, train_loader, val_loader, criterion, optimizer, device, num_epochs=10):

    """
    input:
      model: модель, дообучающаяся на данных
      train_loader: данные для обучения
      val_loader: данные для валидации
      criterion: критерий (функция потерь)
      optimizer: оптимизатор
      device: устройство
      num_epochs: количество эпох обучения (по умолчанию: 5)
    output:
      None
    """

    # Переносим модель на устройство
    model.to(device)

    # Итерируемся по эпохам
    for epoch in range(num_epochs):
        # Режим обучения
        model.train()
        running_loss = 0.0

        for batch in train_loader:
            anchor, positive, negative = batch
            anchor, positive, negative = anchor.to(device), positive.to(device), negative.to(device)

            optimizer.zero_grad()

            # Получаем эмбеддинги
            anchor_embedding = model(anchor)
            positive_embedding = model(positive)
            negative_embedding = model(negative)

            # Вычисляем Triplet Loss
            loss = criterion(anchor_embedding, positive_embedding, negative_embedding)

            # Обратное распространение и оптимизация
            loss.backward()
            optimizer.step()

            running_loss += loss.item()

        # Вычисляем потери и точность на валидации
        model.eval()
        val_loss = 0.0
        val_accuracy = 0.0

        with torch.no_grad():
            for val_batch in val_loader:
                val_anchor, val_positive, val_negative = val_batch
                val_anchor, val_positive, val_negative = val_anchor.to(device), val_positive.to(device), val_negative.to(device)

                # Получаем эмбеддинги для валидации
                val_anchor_embedding = model(val_anchor)
                val_positive_embedding = model(val_positive)
                val_negative_embedding = model(val_negative)

                # Вычисляем Triplet Loss для валидации
                val_loss += criterion(val_anchor_embedding, val_positive_embedding, val_negative_embedding).item()

                # Вычисляем точность для валидации
                val_accuracy += triplet_accuracy(val_anchor_embedding, val_positive_embedding, val_negative_embedding)

        # Подсчитываем точность и потери
        train_loss = running_loss / len(train_loader)
        val_loss /= len(val_loader)
        val_accuracy /= len(val_loader)

        # Выводим информацию об эпохе
        print(f'Epoch {epoch + 1}/{num_epochs}, Train Loss: {train_loss:.4f}, Validation Loss: {val_loss:.4f}, Validation Accuracy: {val_accuracy:.4f}')
```

Функция вычисления Triplet Loss

```python
class TripletLoss(nn.Module):
    def __init__(self, margin=1.0):
        super(TripletLoss, self).__init__()
        self.margin = margin

    def forward(self, anchor, positive, negative):

        # Нормализация эмбеддингов
        anchor_normalized = F.normalize(anchor, p=2, dim=1)
        positive_normalized = F.normalize(positive, p=2, dim=1)
        negative_normalized = F.normalize(negative, p=2, dim=1)

        # Вычисление парных расстояний
        distance_positive = F.pairwise_distance(anchor_normalized, positive_normalized, p=2)
        distance_negative = F.pairwise_distance(anchor_normalized, negative_normalized, p=2)

        # Вычисление Triplet Loss
        loss = F.relu(distance_positive - distance_negative + self.margin)

        # Возвращает потери
        return loss.mean()
```

```python
# Определим критерий обучения
triplet_loss = TripletLoss(margin=0.2)
# Определим оптимизатор
optimizer = optim.Adam(efficientnetv3_triplet_loss.parameters(), lr=0.0005)
# Устройство, на котором происходит обучение
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

**1. Метод распознавания особенностей лица:**

**2. Методы извлечения геометрических особенностей:**

Анализируют геометрические характеристики лица, такие как расстояние между глазами, длина носа, ширина лица и др. Эти параметры могут служить основой для создания уникальных геометрических шаблонов для каждого лица.

**3. Методы геометрической нормализации:**

Нормализуют изображения лиц с учетом их геометрических характеристик, чтобы сделать их более устойчивыми к изменениям в положении, масштабе и ориентации.

**4. Методы поиска точек ориентиров:**

Основаны на поиске ключевых точек, таких как углы глаз, кончики носа, уголки рта. Эти точки могут быть использованы для создания уникальных геометрических шаблонов лиц.

**5. Методы использования геометрических дескрипторов:**

Извлекают геометрические дескрипторы, такие как расстояния между ключевыми точками или углы между определенными частями лица, и используют их для идентификации.

**6. Методы распознавания по форме лица:**

Основаны на анализе формы лица, например, с использованием методов геометрической моделирования.

Эти методы могут применяться как самостоятельно, так и в сочетании с другими техниками распознавания лиц, такими как методы глубокого обучения и нейросетевые подходы. В современных системах распознавания лиц часто применяются методы, использующие комбинацию геометрических и текстурных признаков, что позволяет повысить точность и устойчивость к различным условиям.

*Методы на основе шаблонов и текстур*

Методы на основе шаблонов и текстур для распознавания лиц часто используют информацию о структуре лица и его текстурных характеристиках. Вот несколько методов в этой категории:

**1. Методы использования текстурных признаков:**

Анализируют текстурные характеристики лица, такие как распределение яркости, текстурные шаблоны, особенности морщин и др. Эти признаки могут быть извлечены с использованием различных методов обработки изображений.

**2. Методы локальных бинарных шаблонов (Local Binary Patterns, LBP):**

Основаны на выделении локальных текстурных шаблонов в изображении. LBP описывает текстурные характеристики каждого пикселя, а затем эти описания могут быть использованы для распознавания.

**3. Методы гистограммы ориентированных градиентов (Histogram of Oriented Gradients, HOG):**

Используют градиенты яркости для создания гистограммы направлений градиентов. Это позволяет выделить края и текстурные особенности лица.

**4. Методы Габоровых фильтров:**

Используют Габоровы фильтры для анализа текстурных особенностей, таких как текстуры кожи, морщин и другие.

**5. Методы использования вейвлет-преобразования:**

Применяют вейвлет-преобразование для анализа текстурных деталей на различных масштабах. Это позволяет выделить текстурные характеристики лица.

**6. Методы использования геометрических шаблонов:**

Основаны на выделении геометрических структур, таких как форма глаз, носа и рта. Эти шаблоны могут быть использованы для распознавания лиц.

**7. Методы использования локальных дескрипторов:**

Используют локальные дескрипторы, такие как SIFT (Scale-Invariant Feature Transform) или SURF (Speeded-Up Robust Features), для описания текстурных и структурных особенностей лица.

Эти методы могут использоваться индивидуально или в комбинации с другими техниками, такими как глубокое обучение. Также важно отметить, что современные методы распознавания лиц все чаще используют гибридные подходы, комбинируя информацию о текстуре, геометрии и других признаках для повышения точности и устойчивости к различным условиям.

*Глубокое обучение и нейронные сети для распознавания лиц*

Глубокое обучение и нейронные сети стали ключевыми инструментами в области распознавания лиц благодаря их способности автоматического извлечения признаков из больших объемов данных. Вот несколько ключевых аспектов применения глубокого обучения и нейронных сетей для распознавания лиц:

**1. Сверточные нейронные сети (CNN):**

Сверточные нейронные сети являются мощным инструментом для обработки изображений. Они могут автоматически извлекать иерархию признаков, начиная от низкоуровневых деталей (края, углы) до высокоуровневых абстракций (лица, объекты). Архитектуры, такие как VGG, ResNet и Inception, успешно применяются в задачах распознавания лиц.

**2. Глубокие сети сиамского типа:**

**1. Сети сиамского типа:**


*   В сетях сиамского типа две ветви (например, две нейронные сети) делят параметры между собой. Они обрабатывают пару входных изображений и генерируют их эмбеддинги.
*   Эмбеддинги затем сравниваются, и расстояние между ними используется для определения степени схожести изображений.
*   Функция потерь может быть определена так, чтобы минимизировать расстояние для пар похожих изображений и максимизировать его для пар различных изображений.



Пример обучения нейросети сиамского типа

```python
class SiameseNetwork(nn.Module):
    def __init__(self):
        super(SiameseNetwork, self).__init__()
        self.cnn = models.resnet18(pretrained=True)
        self.fc = nn.Linear(512, 256)

    def forward(self, x1, x2):
        output1 = self.cnn(x1)
        output1 = self.fc(output1)

        output2 = self.cnn(x2)
        output2 = self.fc(output2)

        return output1, output2
```

*   Якорь и положительный пример должны быть похожими (они относятся к одному классу), тогда как отрицательный пример должен быть различным (принадлежать к другому классу).
*   Функция потерь триплетной сети максимизирует расстояние между эмбеддингами якоря и отрицательного примера и минимизирует расстояние между эмбеддингами якоря и положительного примера.

Использование предварительно обученных моделей, таких как ResNet, VGG и других, для извлечения признаков в задаче распознавания лиц является эффективным подходом. Эти модели, предварительно обученные на больших наборах данных, обладают способностью извлекать высокоуровневые признаки из изображений, что может быть полезно в задачах распознавания объектов, включая лица.

Вот пример того, как можно использовать предварительно обученную модель ResNet для извлечения признаков в PyTorch:

```python
# Загружаем предварительно обученную модель ResNet
resnet_model = models.resnet18(pretrained=True)

# Убираем последний слой (классификационный), так как нам нужны признаки, а не предсказания классов
resnet_model = nn.Sequential(*list(resnet_model.children())[:-1])

# Устанавливаем модель в режим оценки (evaluation mode)
resnet_model.eval()

# Пример функции для извлечения признаков из изображения
def extract_features(image_path, model):
    image = Image.open(image_path)

    # Применяем трансформации, соответствующие обучению ResNet
    preprocess = transforms.Compose([
        transforms.Resize(256),
        transforms.CenterCrop(224),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
    ])

    input_tensor = preprocess(image)
    input_batch = input_tensor.unsqueeze(0)  # Добавляем размерность батча

    # Преобразуем вход в переменную PyTorch
    input_var = Variable(input_batch)

    # Передаем изображение через модель (без градиентов)
    with torch.no_grad():
        features = model(input_var)

    # Возвращаем признаки
    return features.squeeze()

# Пример использования
image_path = 'path/to/your/image.jpg'
image_features = extract_features(image_path, resnet_model)

# image_features теперь содержит признаки, извлеченные из изображения с использованием предварительно обученной ResNet
```

*Архитектуры с глубоким вниманием к механизмам внимания*

Архитектуры с глубоким вниманием (deep attention) к механизмам внимания в задаче распознавания лиц стали популярными, так как они позволяют сети акцентировать внимание на важных частях изображения лица. Внимание позволяет моделям сосредотачиваться на определенных областях изображения, улучшая точность распознавания и обработки изображений с более сложными контекстами.

Некоторые из архитектур, использующих механизмы внимания в задаче распознавания лиц:

**1. SENet (Squeeze-and-Excitation Networks):**

Эта архитектура внедряет блоки SE, которые автоматически учат веса для каждого канала входного тензора в зависимости от его содержимого. Это позволяет модели более эффективно использовать информацию из различных частей изображения.



Пример использования SENet

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision.models import resnet50

class SELayer(nn.Module):
    def __init__(self, channel, reduction=16):
        super(SELayer, self).__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(
            nn.Linear(channel, channel // reduction, bias=False),
            nn.ReLU(inplace=True),
            nn.Linear(channel // reduction, channel, bias=False),
            nn.Sigmoid()
        )

    def forward(self, x):
        b, c, _, _ = x.size()
        y = self.avg_pool(x).view(b, c)
        y = self.fc(y).view(b, c, 1, 1)
        return x * y.expand_as(x)

class SENet(nn.Module):
    def __init__(self, num_classes=1000):
        super(SENet, self).__init__()
        self.resnet = resnet50(pretrained=True)
        self.layer0 = nn.Sequential(self.resnet.conv1, self.resnet.bn1, self.resnet.relu, self.resnet.maxpool)
        self.layer1 = self.resnet.layer1
        self.layer2 = self.resnet.layer2
        self.layer3 = self.resnet.layer3
        self.layer4 = self.resnet.layer4
        self.avgpool = self.resnet.avgpool
        self.fc = self.resnet.fc

        # Добавляем SE-модули
        self.se_layer1 = SELayer(256)
        self.se_layer2 = SELayer(512)
        self.se_layer3 = SELayer(1024)
        self.se_layer4 = SELayer(2048)

    def forward(self, x):
        x = self.layer0(x)
        x = self.layer1(x) * self.se_layer1(x)
        x = self.layer2(x) * self.se_layer2(x)
        x = self.layer3(x) * self.se_layer3(x)
        x = self.layer4(x) * self.se_layer4(x)
        x = self.avgpool(x)
        x = x.view(x.size(0), -1)
        x = self.fc(x)
        return x

# Пример использования
input_tensor = torch.randn((1, 3, 224, 224))  # пример входного тензора (батч 1, 3 канала, 224x224)
se_net = SENet()
output = se_net(input_tensor)
print(output.shape)
```

**2. CBAM (Convolutional Block Attention Module):**

CBAM представляет собой блок внимания, который объединяет в себе как канальное, так и пространственное внимание. Он добавляет два параллельных модуля внимания к стандартным сверточным блокам.


Ниже приведен пример использования CBAM в задаче распознавания лиц. Для этого используется модель ResNet с добавлением CBAM-модуля.

```python
class ChannelAttention(nn.Module):
    def __init__(self, in_planes, ratio=16):
        super(ChannelAttention, self).__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.max_pool = nn.AdaptiveMaxPool2d(1)
        self.fc = nn.Sequential(
            nn.Conv2d(in_planes, in_planes // ratio, 1, bias=False),
            nn.ReLU(),
            nn.Conv2d(in_planes // ratio, in_planes, 1, bias=False)
        )

    def forward(self, x):
        avg_pool = self.avg_pool(x)
        max_pool = self.max_pool(x)
        avg_out = self.fc(avg_pool)
        max_out = self.fc(max_pool)
        out = avg_out + max_out
        return torch.sigmoid(out)

class SpatialAttention(nn.Module):
    def __init__(self, kernel_size=7):
        super(SpatialAttention, self).__init__()
        self.conv = nn.Conv2d(2, 1, kernel_size=kernel_size, padding=(kernel_size-1)//2, bias=False)

    def forward(self, x):
        avg_out = torch.mean(x, dim=1, keepdim=True)
        max_out, _ = torch.max(x, dim=1, keepdim=True)
        out = torch.cat([avg_out, max_out], dim=1)
        out = self.conv(out)
        return torch.sigmoid(out)

class CBAM(nn.Module):
    def __init__(self, in_planes, ratio=16, kernel_size=7):
        super(CBAM, self).__init__()
        self.channel_attention = ChannelAttention(in_planes, ratio)
        self.spatial_attention = SpatialAttention(kernel_size)

    def forward(self, x):
        x_out = x * self.channel_attention(x)
        x_out = x_out * self.spatial_attention(x_out)
        return x_out

class ResNetWithCBAM(nn.Module):
    def __init__(self, num_classes=2):
        super(ResNetWithCBAM, self).__init__()
        self.resnet = resnet50(pretrained=True)
        self.cbam1 = CBAM(64)
        self.cbam2 = CBAM(256)
        self.cbam3 = CBAM(512)
        self.cbam4 = CBAM(1024)
        self.cbam5 = CBAM(2048)
        self.fc = nn.Sequential(
            nn.Linear(2048, num_classes),
            nn.Softmax(dim=1)
        )

    def forward(self, x):
        x = self.resnet.conv1(x)
        x = self.resnet.bn1(x)
        x = self.resnet.relu(x)
        x = self.resnet.maxpool(x)

        x = self.resnet.layer1(x)
        x = self.cbam1(x)

        x = self.resnet.layer2(x)
        x = self.cbam2(x)

        x = self.resnet.layer3(x)
        x = self.cbam3(x)

        x = self.resnet.layer4(x)
        x = self.cbam4(x)

        x = self.resnet.avgpool(x)
        x = x.view(x.size(0), -1)
        x = self.fc(x)

        return x
```


**3. Non-local Neural Networks:**

Этот подход представляет блоки с нелинейным глобальным вниманием. Он позволяет моделям моделировать долгосрочные зависимости между пикселями в изображении.


Вот пример использования Non-local Neural Networks в задаче распознавания лиц с использованием библиотеки PyTorch:

```python
class NLNNBlock(nn.Module):
    def __init__(self, in_channels, reduction=2):
        super(NLNNBlock, self).__init__()

        self.inter_channels = max(1, in_channels // reduction)
        self.g = nn.Conv2d(in_channels, self.inter_channels, kernel_size=1)
        self.theta = nn.Conv2d(in_channels, self.inter_channels, kernel_size=1)
        self.phi = nn.Conv2d(in_channels, self.inter_channels, kernel_size=1)
        self.conv_mask = nn.Conv2d(self.inter_channels, in_channels, kernel_size=1)
        self.softmax = nn.Softmax(dim=-1)

    def forward(self, x):
        batch_size, C, H, W = x.size()
        g_x = self.g(x).view(batch_size, self.inter_channels, -1)
        g_x = g_x.permute(0, 2, 1)
        theta_x = self.theta(x).view(batch_size, self.inter_channels, -1)
        phi_x = self.phi(x).view(batch_size, self.inter_channels, -1)

        theta_x = theta_x.permute(0, 2, 1)

        f = torch.matmul(theta_x, phi_x)
        f_div_C = F.softmax(f, dim=-1)
        y = torch.matmul(f_div_C, g_x)
        y = y.permute(0, 2, 1).contiguous()
        y = y.view(batch_size, self.inter_channels, H, W)
        W_y = self.conv_mask(y)
        z = W_y + x

        return z

class ResNetWithNLNN(nn.Module):
    def __init__(self, num_classes=2):
        super(ResNetWithNLNN, self).__init__()
        self.resnet = resnet50(pretrained=True)
        self.nlnn1 = NLNNBlock(64)
        self.nlnn2 = NLNNBlock(256)
        self.nlnn3 = NLNNBlock(512)
        self.nlnn4 = NLNNBlock(1024)
        self.nlnn5 = NLNNBlock(2048)
        self.fc = nn.Sequential(
            nn.Linear(2048, num_classes),
            nn.Softmax(dim=1)
        )

    def forward(self, x):
        x = self.resnet.conv1(x)
        x = self.resnet.bn1(x)
        x = self.resnet.relu(x)
        x = self.resnet.maxpool(x)

        x = self.resnet.layer1(x)
        x = self.nlnn1(x)

        x = self.resnet.layer2(x)
        x = self.nlnn2(x)

        x = self.resnet.layer3(x)
        x = self.nlnn3(x)

        x = self.resnet.layer4(x)
        x = self.nlnn4(x)

        x = self.resnet.avgpool(x)
        x = x.view(x.size(0), -1)
        x = self.fc(x)

        return x
```
